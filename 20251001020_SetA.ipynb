# Que 1 Equilibrium Conversion Checker
For the liquid-phase equilibrium reaction:

A + B ⇌ C

Initial concentrations are:

[A]0 = 1.0 mol/L
[B]0 = 1.0 mol/L
[C]0 = 0 mol/L

If x mol/L of C is formed:

Kc = x / (1 - x)^2

This gives the quadratic equation:

Kc*x^2 - (2*Kc + 1)*x + Kc = 0

Write a simple beginner-level Python program that:

1. Reads Kc using input.
2. Prints the input type and converted value type.
3. Displays a clear message if Kc is not a number or Kc <= 0.
4. Calculates a, b, c, the discriminant, and both roots.
5. Selects the physically meaningful root where 0 <= x <= 1.
6. Calculates and prints equilibrium [A], [B], and [C] to 3 decimal places.
7. Calculates conversion as x/[A]0 * 100.
8. Classifies conversion: greater than 70%: high, 40% to 70%: moderate, below 40%: low


Write the python program from a second year chemical engineering student's point of view



```python
Kc = input("Enter Kc: ")

print("Input type:", type(Kc))

try:
    Kc = float(Kc)

    print("Converted Kc:", Kc)
    print("Converted type:", type(Kc))

    if Kc <= 0:
        print("Invalid input: Kc must be greater than zero.")
    else:
        a = Kc
        b = -(2 * Kc + 1)
        c = Kc

        discriminant = b**2 - 4 * a * c

        print("a =", a)
        print("b =", b)
        print("c =", c)
        print("Discriminant =", discriminant)

        if discriminant < 0:
            print("No real roots exist.")
        else:
            root1 = (-b + discriminant**0.5) / (2 * a)
            root2 = (-b - discriminant**0.5) / (2 * a)

            print("Root 1 =", root1)
            print("Root 2 =", root2)

            if 0 <= root1 <= 1:
                x = root1
            elif 0 <= root2 <= 1:
                x = root2
            else:
                x = None

            if x is None:
                print("No physically meaningful root.")
            else:
                A = 1 - x
                B = 1 - x
                C = x

                conversion = x * 100

                print(f"[A] = {A:.3f} mol/L")
                print(f"[B] = {B:.3f} mol/L")
                print(f"[C] = {C:.3f} mol/L")
                print(f"Conversion = {conversion:.3f}%")

                if conversion > 70:
                    print("Conversion classification: high")
                elif conversion >= 40:
                    print("Conversion classification: moderate")
                else:
                    print("Conversion classification: low")

except ValueError:
    print("Invalid input: Kc must be a number.")
```

# Que 2 Batch reactor stimulation and batch tracking Part A
 
Consider the first-order reaction:

A → Products

The rate equation is:

dC/dt = -kC

Use these given values:

 Initial concentration, C0 = 2.0 mol/L
 Rate constant, k = 0.08 min^-1
 Time step, Δt = 1 min
 Simulation time = 10 min

Use the Euler method:

C_new = C_old - k × C_old × Δt

Write a simple Python program that:

1. Uses a for loop to simulate the reaction for 10 minutes.
2. Prints time, concentration, and conversion at each minute.
3. Calculates conversion using:
   Conversion (%) = ((C0 - C) / C0) × 100
4. Uses math.exp() to calculate the exact concentration at 10 minutes.
5. Calculates the absolute error between the Euler and exact results.
6. Uses a while loop to continue the simulation until conversion reaches 60% or time reaches 100 minutes.
7. Prints the time required and final conversion.

Write the python program from a second year chemical engineering student's point of view

```python
C0 = 2.0
k = 0.08
dt = 1.0

C = C0

print("Time (min)\tConcentration (mol/L)\tConversion (%)")

for time in range(11):
    conversion = ((C0 - C) / C0) * 100

    print(f"{time}\t\t{C:.4f}\t\t\t{conversion:.2f}")

    if time < 10:
        C = C - k * C * dt
```

```python
import math

C0 = 2.0
k = 0.08
dt = 1.0

C_euler = C0

for time in range(10):
    C_euler = C_euler - k * C_euler * dt

C_exact = C0 * math.exp(-k * 10)
absolute_error = abs(C_euler - C_exact)

print("Euler concentration:", round(C_euler, 4), "mol/L")
print("Exact concentration:", round(C_exact, 4), "mol/L")
print("Absolute error:", round(absolute_error, 4), "mol/L")
```

```python
C0 = 2.0
k = 0.08
dt = 1.0

C = C0
time = 0
conversion = 0

while conversion < 60 and time < 100:
    C = C - k * C * dt
    time = time + 1
    conversion = ((C0 - C) / C0) * 100

print("Time needed:", time, "minutes")
print("Final concentration:", round(C, 4), "mol/L")
print("Final conversion:", round(conversion, 2), "%")
```

# Part B 
I am a second year chemical engineering student learning Python

Batch labels such as PLANT20261030 are stored as text

Write a simple beginner level Python program that

1 Reads batch labels as text
2 Loops through every character in each label
3 Counts the number of digits
4 Calculates the sum of the digits
5 Displays a clear message if the label is empty
6 Displays a clear message if the label contains no digits
7 Tests the program with PLANT2026103 an empty label and ABCDEF

Use simple Python suitable for a second year chemical engineering student

```python
# Batch-label digit checker
# Chemical engineering programming practice

labels = ["PLANT20261030", "", "ABCDEF"]

for label in labels:
    print("Batch label:", label)

    if label == "":
        print("Message: The label is empty.")
    else:
        digit_count = 0
        digit_sum = 0

        for character in label:
            if character.isdigit():
                digit_count = digit_count + 1
                digit_sum = digit_sum + int(character)

        if digit_count == 0:
            print("Message: The label contains no digits.")
        else:
            print("Number of digits:", digit_count)
            print("Sum of digits:", digit_sum)
```

# que 3 part a

A laboratory has n candidate additives and wants to test r additives together in one formulation. The order of selecting additives does not matter.

Write a simple Python program that:

1. Defines a recursive factorial function.
2. Uses the base case factorial(0) = 1.
3. Defines a combination function using:
   combination(n, r) = n! / (r! * (n-r)!)
4. Rejects invalid input when n or r is negative or r is greater than n.
5. Calculates the number of possible formulations for n = 6 and r = 3.
6. Displays a clear invalid-input message for n = 5 and r = 7.

Use beginner-level Python suitable for a second-year chemical engineering student.

```python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    if n == 0:
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if n < 0 or r < 0 or r > n:
        print("Invalid input: 0 <= r <= n is required.")
        return None

    return factorial(n) // (factorial(r) * factorial(n - r))


n = 6
r = 3

result = combination(n, r)
print("Number of possible formulations:", result)


n = 5
r = 7

combination(n, r)
```

# que 3 part B
During a reaction, samples are collected at times given by Fibonacci numbers multiplied by 2 minutes.

Write a simple Python program that:

1. Defines an iterative function fib_iter(n) that returns the first n Fibonacci numbers.
2. Defines a recursive function fib_rec(k) that returns the kth Fibonacci number.
3. Uses F0 = 0 and F1 = 1.
4. Uses n = 10.
5. Checks whether the iterative and recursive results agree.
6. Prints a message confirming whether both functions agree.
7. Prints a table with:
   - Sample number
   - Fibonacci tag
   - Sampling time in minutes
8. Calculates sampling time as:
   Sampling time = Fibonacci number × 2

Use beginner-level Python suitable for a second-year chemical engineering student.

```python
def fib_iter(n):
    numbers = []
    a = 0
    b = 1

    for i in range(n):
        numbers.append(a)
        a, b = b, a + b

    return numbers


def fib_rec(k):
    if k == 0:
        return 0
    elif k == 1:
        return 1
    else:
        return fib_rec(k - 1) + fib_rec(k - 2)


n = 10

iterative_numbers = fib_iter(n)
recursive_numbers = []

for k in range(n):
    recursive_numbers.append(fib_rec(k))

if iterative_numbers == recursive_numbers:
    print("Both Fibonacci functions agree.")
else:
    print("The Fibonacci functions do not agree.")

print()
print("Sample number\tFibonacci tag\tSampling time (min)")

for sample in range(n):
    fibonacci_tag = iterative_numbers[sample]
    sampling_time = fibonacci_tag * 2

    print(f"{sample + 1}\t\t{fibonacci_tag}\t\t{sampling_time}")
```

# que 4 (d)

The experimental data is given in the following table:

| Volume (L) | Pressure (bar) |
|---:|---:|
| 1 | 4.00 |
| 2 | 2.03 |
| 3 | 1.31 |
| 4 | 1.02 |
| 5 | 0.81 |

Write a simple Python program that:

1. Stores the volume and pressure values.
2. Calculates P × V for every row.
3. Calculates k as the mean of the P × V values.
4. Uses the reference equation P = k / V.
5. Plots pressure against volume.
6. Plots the experimental data and the reference curve on the same graph.
7. Adds a title.
8. Adds axis labels with units.
9. Adds a legend and grid.

Use beginner-level Python suitable for a second-year chemical engineering student.

```python
import numpy as np
import matplotlib.pyplot as plt

# Experimental data
volume = [1, 2, 3, 4, 5]
pressure = [4.00, 2.03, 1.31, 1.02, 0.81]

# Calculate PV values
pv_values = []

for i in range(len(volume)):
    pv = pressure[i] * volume[i]
    pv_values.append(pv)

# Calculate the mean value of k
k = sum(pv_values) / len(pv_values)

print("PV values:", pv_values)
print("Mean value of k:", k)

# Create values for the reference curve P = k / V
volume_curve = np.linspace(1, 5, 100)
pressure_curve = k / volume_curve

# Plot experimental data
plt.plot(
    volume,
    pressure,
    "o",
    label="Experimental data"
)

# Plot reference curve
plt.plot(
    volume_curve,
    pressure_curve,
    "-",
    label="Reference curve: P = k/V"
)

plt.title("Pressure versus Volume")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True)

plt.show()
```
